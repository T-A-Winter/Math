### Aufgabe 1

Für welche Werte des Parameter $\beta$ hat das lineare Optimierungsproblem:

$ f(x,y)= x + \beta y  \rightarrow    \text{max(min)}$

unter den Nebenbedingungen:

$x  - y\leq 1,\quad x + 2y \geq 4,\quad x \geq 0, \quad y \geq 0$

a) eine einzige Lösung;

b) unendlich viele Lösungen;

c) keine Lösung.


```python
import numpy as np
from scipy.optimize import linprog

# Zielfunktion
c = np.array([0, 0, 0, 0])

# Die Koeffizientenmatrix der Nebenbedingungen
A = np.array([[1, -1, 1, 0], [-1, -2, 0, 1]])

# Die rechte Seite der Nebenbedingungen
b = np.array([1, -4])

# Setze die Grenzen für die Variablen x, y, s1 und s2
bounds = [(0, None), (0, None), (0, None), (0, None)]

# Löse das lineare Optimierungsproblem
#method='simplex' -> deprecated 
result = linprog(c, A_eq=A, b_eq=b, bounds=bounds, method='highs')

# Ausgabe der Ergebnisse
print("Eckpunkte der zulässigen Region:", result.x)
```

### Aufgabe 4.

Für welche der folgenden Tripel $(x,y,z)$ von Punkten ist $z$ Konvexkombination von $x$ und $y$:

a) $x = (1,1)^T, \, y = (-8, -8)^T,\, z= (3,3)^T$

b) $x = (7,4)^T, \, y = (5.5, 1)^T, \,z= (6,2)^T$

c) $x = (1,1,1)^T,  \, y = (5,9,13)^T, \,z= (2,3,4)^T$

Lösung:

Folgende Definition ist gegeben: 

(Kopie aus Tutorium Seite 3)

Seien x und y Punkte und λ ∈ [0, 1], dann bezeichnen
wir z := λx + (1 − λ)y als eine Konvexkombination von x und y

Es folgt, dass $\lambda$ nicht negativ sein darf und kleiner als 1 sein muss.
Das kann auf folgende art überprüft werden:

1) Stelle das Gleichungssystem nach der Definition auf.
2) Löse die erste gleichung nach $\lambda$ auf und setzte in die übrigen ein. 

Es reicht also nur die erste Gleichung aus dem Gleichungssystem aufzustellen und zu überpüfen ob $\lambda$ zwischen 0 und 1 ist. Wenn $\lambda$ zwischen 0 und 1 ist, dann liegt z auf der Strecke zwischen x und y und es handelt sich um eine Konvexkombination.

```python
def find_lambda(x, y, z):
    #Gleichung nach Lambda aufgelöst
    return (z-y)/(x-y)


# Nur die ersten Gleichungen für die Berechnung von Lambda sind nötig
x_1 = (1,7,1)
y_1 = (-8,5.5,5)
z_1 = (3,6,2)

# Berechnung von λ
resForLambda = [] 
for i in range(len(x_1)):
    resForLambda.append(find_lambda(x_1[i], y_1[i], z_1[i]))
print(resForLambda)
```

Wie sich zeigt ist nur a) keine Konvexkombination von x und y da $\lambda$ hier $\geq$ 1 ist
