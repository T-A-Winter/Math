### Aufgabe 2


 a) Geben Sie ein Optimierungsproblem an, das zwar zulässige Lösungen besitzt, aber keine optimale Lösung.
 
$$
    ZF: x + 2y \rightarrow min! \\
    NB: 3x \geq 6 \\
        y \geq 0
$$

```python
ecken = np.array([[0,2]])
delta = 1
x = np.arange(-1, 12, delta)
y = np.arange(-1, 12, delta)
X, Y = np.meshgrid(x, y)

p = np.array([1, 2])
Z = p[0]*X + p[1]*Y

fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(111)
ax.add_patch(plt.Polygon(ecken, closed=True, color='green', alpha=0.3))
plt.arrow(0, 0, p[0], p[1], head_width=0.2, fc='k')
CS = plt.contour(X, Y, Z, levels= np.linspace(0, 30, 6), cmap= 'coolwarm')
plt.clabel(CS, inline=1, fontsize=12, fmt='%1.1f')
plt.xlim(-1,12.5)
plt.ylim(-1,12.5)
plt.xlabel('$x$')
plt.ylabel('$Y$')
plt.grid(True)
```

### Aufgabe 4

Lösen Sie graphisch:

$ 3 x + y  \rightarrow   \text{max}$

$2x + y  \geq   2 , \quad 3x + 4y   \leq  12 ,  \quad   x \leq   3,   \quad  x, \, y  \geq  0 $

Anleitung: Man ermittle graphisch die zulässige Menge S in der $(x,y)$ - Ebene, und zeichne die Geraden   $ 3 \, x  + y = c$ (ZF = konstant) für einige Werte $c$.

Lösung:

NB nach y auflösen und in einen Graphen einzeichnen.

NB 1
$$
    2x + y  \geq   2\\
    = y \geq 2 - 2x \\
$$
NB 2
$$
    3x + 4y   \leq  12\\
    = 4y \leq 12 - 3x \\
    = y \leq 3 - \frac{3x}{4} \\
$$

NB 3

$$
     x \leq   3 
$$


```python
ecken = np.array([[4,0],
               [0,3],
               [0,2],
               [1,0]])
delta = 1
x = np.arange(-1, 12, delta)
y = np.arange(-1, 12, delta)
X, Y = np.meshgrid(x, y)

p = np.array([3, 1])
Z = p[0]*X + p[1]*Y

fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(111)
ax.add_patch(plt.Polygon(ecken, closed=True, color='green', alpha=0.3))
plt.arrow(0, 0, p[0], p[1], head_width=0.2, fc='k')
CS = plt.contour(X, Y, Z, levels= np.linspace(0, 30, 6), cmap= 'coolwarm')
plt.clabel(CS, inline=1, fontsize=12, fmt='%1.1f')
plt.xlim(-1,12.5)
plt.ylim(-1,12.5)
plt.xlabel('$x$')
plt.ylabel('$Y$')
plt.grid(True)
```

### Aufgabe 5

Lösen Sie graphisch:

$ x + y  \rightarrow    \text{max}$

$3x + y\leq 8,\quad y + z \leq 5,\quad x + 2z = 2, \quad x, y, z \geq 0$

Anleitung:  Man  drücke $z$ durch $x$ aus und formuliere die Bedingung $ z \geq 0 $ in eine Bedingung an $x$ um. Weiter wie bei Beispiel 4. 


Lösung: 

Aus $ x + y = 0$ folgt $y = -x$

NB1:
$$
    3x + y\leq 8\\
    = 3x + -x\leq 8\\
    = 2x \leq 8\\
    x \leq 4 \\
$$

NB2:
$$
    y + z \leq 5\\
    = -x + z \leq 5 \\
    z \leq 5 + x
$$

NB3:

$$
    x + 2z = 2 \\ 
    2z = 2 -x \\
    z = 1 - \frac{x}{2}
$$

```python
ecken = np.array([[2,0],
               [4,0],
               [9,4],
               [-(8/3),(7/3)]])
delta = 1
x = np.arange(-1, 12, delta)
y = np.arange(-1, 12, delta)
X, Y = np.meshgrid(x, y)

p = np.array([1, 1])
Z = p[0]*X + p[1]*Y
m = np.array([9,4])

fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(111)
ax.add_patch(plt.Polygon(ecken, closed=True, color='green', alpha=0.3))
plt.arrow(0, 0, p[0], p[1], head_width=0.2, fc='k')
plt.arrow(0, 0, m[0], m[1], head_width=0.2, fc='k')
CS = plt.contour(X, Y, Z, levels= np.linspace(0, 30, 15), cmap= 'coolwarm')
plt.clabel(CS, inline=1, fontsize=12, fmt='%1.1f')
plt.xlim(-1,12.5)
plt.ylim(-1,12.5)
plt.xlabel('$x$')
plt.ylabel('$Y$')
plt.grid(True)
```
