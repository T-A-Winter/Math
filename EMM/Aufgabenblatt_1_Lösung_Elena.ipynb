### Aufgabe 4

Lösen Sie graphisch:

$ 3 x + y  \rightarrow   \text{max}$

$2x + y  \geq   2 , \quad 3x + 4y   \leq  12 ,  \quad   x \leq   3,   \quad  x, \, y  \geq  0 $

Anleitung: Man ermittle graphisch die zulässige Menge S in der $(x,y)$ - Ebene, und zeichne die Geraden   $ 3 \, x  + y = c$ (ZF = konstant) für einige Werte $c$.

Umformen nach y:

$2x + y \geq 2 \\
y \geq = -2x + 2$
<br>
$3x + 4y   \leq  12 \\
y \leq 3 - \frac{3}{4}x$
<br>
$3x + y = c \\
y = -3x + c$

```python
import numpy as np
from matplotlib import pyplot as plt
import math

plt.plot(x, -2*x+2, color="black",label = "nb1") 
plt.plot(x, -3/4*x+3, color="blue", label = "nb2")
plt.vlines(3,0,4,color="violet", label = "nb3")

x = np.linspace(0,4,500)
y = np.linspace(0,4,500)
x1,y1 = np.meshgrid(x,y)

nb1 = (2 * x1 + y1 >=2)
nb2 = (3 * x1 + 4 * y1 <=12)
nb3 = (x1 <= 3)
nb4 = (y1 >= 0)
nb5 = (x1 >= 0)

plt.imshow((nb1 & nb2 & nb3 & nb4 & nb5).astype(float), extent = (x1.min(),x1.max(),y1.min(),y1.max()), origin="lower", cmap="Greys", alpha = 0.1);

for c in range(3,12,2):
    plt.plot(x, -3*x + c,"-", color="red")

plt.plot(3,0.75, "o", color = "green",label = "max")
plt.xlim(0,4)
plt.ylim(0,4)
plt.legend()
plt.show()
```

### Aufgabe 5

Lösen Sie graphisch:

$ x + y  \rightarrow    \text{max}$

$3x + y\leq 8,\quad y + z \leq 5,\quad x + 2z = 2, \quad x, y, z \geq 0$

Anleitung:  Man  drücke $z$ durch $x$ aus und formuliere die Bedingung $ z \geq 0 $ in eine Bedingung an $x$ um. Weiter wie bei Beispiel 4. 


$y \leq -3x + 8$ <br>
$y \leq -z + 5 \quad $ -> $ \quad y \leq - \left(1 - \dfrac{1}{2}x\right) + 5$ <br>
$z = 1 - \dfrac{1}{2}x \quad$ -> $ \quad x+2*(1 - \dfrac{1}{2}x) = 2 \quad$ -> $ \quad2=2$<br>
$z \geq 0 \quad -> \quad 1 - \dfrac{1}{2}x \geq 0$

```python
import numpy as np
from matplotlib import pyplot as plt
import math

plt.plot(x, -3*x+8, color="black",label = "nb1") 
plt.plot(x, -(1-(1/2)*x)+5, color="violet", label = "nb2")
plt.vlines(2,0,8,color="green", label = "nb3")

x = np.linspace(0,9,500)
y = np.linspace(0,9,500)
x1,y1 = np.meshgrid(x,y)

nb1 = (3*x1 + y1 <= 8)
nb2 = (y1 + (1-(1/2)*x1) <= 5)
nb3 = (y1 >= 0)
nb4 = (x1 >= 0)
nb5 = ((1-(1/2)*x1) >= 0)

plt.imshow((nb1 & nb2 & nb3 & nb4 & nb5).astype(float),extent = (x1.min(),x1.max(),y1.min(),y1.max()), origin="lower", cmap="Greys", alpha = 0.1);

for c in range(3,8,2):
    plt.plot(x, c - x,"-", color="red")

plt.plot(1.14,4.6, "o", color = "forestgreen",label = "max")

plt.xlim(0,8)
plt.ylim(0,8)
plt.legend()
plt.show()
```
