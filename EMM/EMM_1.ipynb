### Aufgabe 5

Lösen Sie graphisch:

$ 3 x + y  \rightarrow   \text{max}$

$2x + y  \geq   2 , \quad 3x + 4y   \leq  12 ,  \quad   x \leq   3,   \quad  x, \, y  \geq  0 $

Anleitung: Man ermittle graphisch die zulässige Menge S in der $(x,y)$ - Ebene, und zeichne die Geraden   $ 3 \, x  + y = c$ (ZF = konstant) für einige Werte $c$.


```python
from scipy.optimize import linprog

#NB (<=)
#NB l
b = [-2, 12, 3]
#NB r 
A = [[-2, -1], [3, 4], [1, 0]]
#ZF (max)
c = [-3, -1]
model_lin = linprog(c, A, b)
print("Solution:", model_lin.x , "\nfunction val:",model_lin.fun*-1)
#print("\nfull:\n",model_lin)
```

```python
# Copied from moodle tutorial: https://moodle.univie.ac.at/pluginfile.php/22800305/mod_resource/content/1/Lineares_Optimierungsproblem_Visualisieren.ipynb
# and adapted
import numpy as np
from matplotlib import pyplot as plt

# x, y from linprog
rx1 = 3
rx2 = 0.75
rc = 9.75

def drawSolution(rx1, rx2):
    plt.arrow(rx1, 0, 0, maxX, head_width=0.2, fc='purple', label=rx1)
    plt.arrow(0, rx2, maxY, 0, head_width=0.2, fc='purple', label=rx2)

def drawArrow(maxX, maxY, c):
    plt.arrow(0, 0, c[0], c[1], head_width=0.2, fc='k')

def drawLines(maxX, maxY, c):
    delta = 1
    x = np.arange(0, maxX, delta)
    y = np.arange(0, maxY, delta)
    X, Y = np.meshgrid(x, y)
    Z = c[0] * X + c[1] * Y
    CS = plt.contour(X, Y, Z, levels=np.linspace(0, rc, 5), cmap='coolwarm')
    plt.clabel(CS, inline=1, fontsize=12, fmt='%1.1f')
    
    
def fillRectangle(ax, x, y, width, height):
    ax.add_patch(plt.Rectangle((x, y), width, height, facecolor="green", alpha=0.3))
    
def drawInequality(inequality, ax, maxX, maxY):
    delta = 0.1
    x = np.arange(0, maxX, delta)
    y = np.arange(0, maxY, delta)
    for xi in x:
        for yi in y:
            if inequality(xi, yi):
                fillRectangle(ax, xi, yi, delta, delta)
    
def finishPlot(maxX, maxY):
    plt.xlim(0, maxX)
    plt.ylim(0, maxY)
    plt.xlabel('$x$')
    plt.ylabel('$Y$')
    plt.grid(True)
    plt.show()
    
maxX = 10
maxY = 10

c = np.array([3, 1])
fig = plt.figure(figsize = (8, 8))
ax = fig.add_subplot(1, 1, 1)

drawArrow(maxX, maxY, c)
drawLines(maxX, maxY, c)

inequality = lambda x, y : 2*x + y >= 2
inequality2 = lambda x, y: 3*x + 4*y <= 12
inequality3 = lambda x, y : x<=3
inequality4 = lambda x, y : (x>=0, y>=0)
drawInequality(inequality, ax, maxX, maxY)
drawInequality(inequality2, ax, maxX, maxY)
drawInequality(inequality3, ax, maxX, maxY)
#drawInequality(inequality4, ax, maxX, maxY)
drawSolution(rx1, rx2)
finishPlot(maxX, maxY)
    
```
