### Aufgabe 1

Wie viele gleich lange Worte lassen sich aus dem Wort STATISTIK bilden?

Es handelt sich um das Model "Ziehen ohne zurücklegen unter berücksichtigung der Anordnung"

Die Formel:

N* = Anzahl der unterscheidbaren Stichproben

N = Umfang der Grundgesamtheit

n = Anzahl der Elemente in der Stichprobe

$$
    N* = \frac{N!}{(N-n)!}
$$

Zu berücksichtigen $\rightarrow$ S und I kommen je 2 mal vor und T 3 mal.

N = 9 $\rightarrow$ Länge
n = 5 $\rightarrow$ anzahl der Buchstaben 

$$
    \frac{9!}{(9-5)!} = 15120
$$

```r
factorial(9)/factorial(9-5)
```

### Aufgabe 2

Einer Studentin werden bei einer Prüfung 12 Fragen vorgelegt. Sie darf davon 4 auswählen. Man berechne die Anzahl der Wahlmöglichkeiten, wenn sie aus den ersten 6 Fragen mindestens drei auswählen muss.

Es ist das Model "Ziehen ohne zurücklegen ohne berücksichtigung der Anordnung"

Formel:

$$
    N* = \left( \begin{array}{c}N \\ n \end{array}\right)
$$

Es gibt zwei Möglichkeiten wie sich die Studendtin entscheiden kann. Entweder sie wählt aus den ersten 6 Fragen 3 aus und 1 aus anderen 6. Oder sie wählt alle 4 Fragen aus den ersten 6.

$$
        N* =\left( \begin{array}{c} 6 \\ 3 \end{array}\right) \cdot
             \left( \begin{array}{c} 6 \\ 1 \end{array}\right) +
             \left( \begin{array}{c} 6 \\ 4 \end{array}\right) = 135
$$

```r
choose(6,3)*choose(6,1) + choose(6,4)
```

### Aufgabe 3

Bei einem Radrennen mit 10 Startern wird in jeweils 4 Rennen der
Sieger ermittelt. Wie viele mögliche Gesamtergebnisse gibt es, wenn
nur die Anzahl der gewonnenen Rennen berücksichtigt wird?

Es handelt sich hier um das Model "Ziehen mit zurücklegen ohne berücksichtigung der Anordnung"

Formel:

$$
    N* = \left( \begin{array}{c}N + n -1 \\ n \end{array}\right)
$$

N = 10 Radrenner
n = 1 Sieger 

$$
    N* = \left( \begin{array}{c} 10 + 4 -1 \\ 4 \end{array}\right) = 715
$$



```r
choose(10 + 4 - 1, 4)
```

### Aufgabe 4

Aus 5 Ehepaaren werden zufällig 4 Personen ausgewählt. Wie viele unterschiedliche Anordnungen gibt es, wenn sich unter den 4 ausgewählten Personen kein Ehepaar befinden soll?

LÖSUNG:

Es gibt $\left( \begin{array}{c} 10 \\ 1 \end{array}\right)$ Möglichkeiten 1 Person aus 10 zu wählen.

Wir die zweite Person gibt es nur noch 8 Möglichkeiten diese zu wählen. (Die zweite Person darf nicht der Partner der esten Person sein $\rightarrow$ 9 Personen - 1 Person, welche der Partner ist)

Für die 3 Personen gibt es dann nur noch 6 Möglichkeiten. (Die 3. Person darf kein Partner der bereits gezogenen Personen sein)

usw. 

Es kommt zur rechnung:

$$
    10 \cdot 8 \cdot 6 \cdot 4 = 1920
$$

```r
10*8*6*4
```

### Aufgabe 6

Man berechne die Anzahl der vierstelligen Zahlen, welche die Ziffern 1 und 3 enthalten, wenn die Zahlen aus unterschiedlichen Ziffern bestehen.

Da 1 und 3 für uns schon gewählt wurden, kann man nur noch aus 8 Ziffern eine wählen $\rightarrow \left(\begin{array}{c} 8 \\ 1\end{array}\right)$.

Für die 2. Ziffer bleiben nur noch 7 Möglichkeiten übrig $\rightarrow \left(\begin{array}{c} 7 \\ 1\end{array}\right)$.

Die 3. und die 4. müssen nicht mehr gewählt werden. Sie werden nur noch in den leeren Plätzen platziert. 

Da die Reihenfolge der Ziffern egal ist muss man mit $4!$ multiplizieren um alle Möglichkeiten zu beachten.

```r
choose(8,1) * choose(7,1) * factorial(4)
```
