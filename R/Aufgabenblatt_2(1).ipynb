### Aufgabe 1

Neun Personen besteigen einen Zug mit drei Wagen. Jede Person wählt zufällig und unabhängig von den anderen Personen einen Wagen. Wie groß ist unter geeigneter Laplace-Annahme die Wahrscheinlichkeit dafür, dass

a) genau drei Personen in den ersten Wagen steigen,  
b) jeweils drei Personen in jeden Wagen steigen,  
c) die neun Personen sich in Gruppen zu zwei, drei und vier Personen auf die drei Wagen aufteilen?

Bernoulli Experiment: Es kann also die Formel
$$\\ \left( \begin{array}{c} n \\ k \end{array}\right) \cdot  p^k(1-p)^{n-k}$$
 
genommen werden.

a) n = 9; k = 3; p = 1/3

$$\\ \left( \begin{array}{c} 9 \\ 3 \end{array}\right) \cdot  \frac{1}{3}^3(1-\frac{1}{3})^{9-3}$$

b)$$\\$$
Wahrscheinlichkeit dafür das genau 3 Personen in den ersten Wagen einsteigen MAL der Wahrscheinlichkeit das genau 3 Personen von den übrigen 6 in den zweiten Wagen einsteigen.
$$\\ \left( \begin{array}{c} 9 \\ 3 \end{array}\right) \cdot  \frac{1}{3}^3(1-\frac{1}{3})^{9-3} \cdot 
 \left( \begin{array}{c} 6 \\ 3 \end{array}\right) \cdot  \frac{1}{2}^3(1-\frac{1}{3})^{6-3}$$
 
c)
Wahrscheinlichkeit das genau 2 Personen in den ersten Wagen steigen MAL der Wahrscheinlichkeit das genau 3 von 7 Personen in den zweiten Wagen steigen.
$$\left( \begin{array}{c} 9 \\ 2 \end{array}\right) \cdot  \frac{1}{3}^2(1-\frac{1}{3})^{9-2} \cdot
\left( \begin{array}{c} 7 \\ 3 \end{array}\right) \cdot  \frac{1}{2}^3(1-\frac{1}{2})^{7-3}$$

```r
choose(9,3)*(1/3)^3*(1-(1/3))^6
```

```r
(choose(9,3)*(1/3)^3*(1-(1/3))^6)*(choose(6,3)*(1/2)^3*(1-(1/2))^3)
```

```r
(choose(9,2)*(1/3)^2*(1-(1/3))^7)*(choose(7,3)*(1/2)^3*(1-(1/2))^4)
```

### Aufgabe 2

Aus einer Tabelle (einstelliger) Zufallsziffern werden nacheinander Ziffern entnommen. Wie viele Ziffern muss man mindestens entnehmen, damit unter ihnen mit einer Wahrscheinlichkeit $\geq 0.95$ mindestens eine ungerade ist.

$$
    1-(\frac{1}{2})^n \geq 0.95 \\
    -\frac{1}{2}^n \geq 0.95 - 1 \\
    \frac{1}{2}^n \leq 0.05 \\ 
    n \cdot \ln{\frac{1}{2}} \leq \ln{0.05} \\
    n \geq \frac{\ln{0.05}}{\ln{\frac{1}{2}}}
$$
mindestens 5 Ziffern

```r
log(0.05)/log(1/2)
```

$$\\ \left( \begin{array}{c} n \\ k \end{array}\right) \cdot  p^k(1-p)^{n-k}\\

### Aufgabe 6

Beim "Doppelkopf-Spiel" wird ein Kartenspiel mit 48 Karten durchgemischt und es werden an vier Spieler je zwölf Karten verteilt. Acht der 48 Karten heißen "Damen", zwei der acht Damen heißen "Kreuz-Damen". Man berechne unter der Laplace-Annahme die Wahrscheinlichkeit für folgendes Ereignis: Ein Spieler erhält beide Kreuz-Damen.

Überlegung:
4 Spieler, es ist egal bei wem das Ereigniss passiert. 
$$
    \\
$$
(Ein Spieler bekommt 2 Kreuzdamen) / (Gesammt möglichkeit) mal 4 Spieler

$$
    4 \cdot \frac{\left( \begin{array}{c} 46 \\ 10 \end{array}\right)}
    {\left( \begin{array}{c} 48 \\ 12 \end{array}\right)}
$$

```r
4*((choose(46,10) / choose(48,12)))
```
