## Laplace-Wahrscheinlichkeiten

### Aufgabe 1

Neun Personen besteigen einen Zug mit drei Wagen. Jede Person wählt zufällig und unabhängig von den anderen Personen einen Wagen. Wie groß ist unter geeigneter Laplace-Annahme die Wahrscheinlichkeit dafür, dass

a) genau drei Personen in den ersten Wagen steigen,  
b) jeweils drei Personen in jeden Wagen steigen,  
c) die neun Personen sich in Gruppen zu zwei, drei und vier Personen auf die drei Wagen aufteilen?

Für Laplace Zufallsexperimenten ist jeder Versuchsausgang gleich wahrscheinlich. Die Wahrscheinlichkeit für jeden Versuchsausgang ist gleich wahrscheinlich. Die Wahrscheinlichkeit P für ein bestimmte Ereignis A ist gleich der Anzahl der günstigen Versuchsausgänge (bei dem das Ereignis A auftritt) durch die Anzahl aller möglichen Versuchsausgängen.

$$\Large\,\, P(A) \,\, = \,\,  \frac{Anzahl \,\, aller \,\, günstigen \,\,Versuchsausgänge\,\,}{Anzahl \,\,aller \,\,möglichen \,\,Versuchsausgänge\,\,} $$

Wir haben 3 verschiedene Zustände (1, 2 oder 3) und 9 Personen. Die Anzahl der möglichen Permutationen ,welche angibt wie jeder Wagen bestiegen ist beträgt $\Large 3^9 \,\, = \,\, 19683$. Dies entspricht einer Matrix mit 19683 Zeilen und 9 Spalten.

```r
3**9 
```

3 der 9 Personen besteigen den ersten Wagen. Die Anzahl der Möglichkeiten welche der 9 Personen den ersten Wagen belegt wird durch den Binomialkoeffizienten $ \left( \begin{array}{c} 9 \\ 3 \end{array} \right) $ bestimmt. Die restlichen 6 Personen können sich auf die 2 restlichen Wagons aufteilen. Die Anzahl der Möglichkeiten dafür $2^6$.

2**6

Insgesamt berechnet sich die Wahrscheinlichkeit folgendermaßen:
$$\Large \frac{\left( \begin{array}{c} 9 \\ 3 \end{array}\right) \,\, \cdot \,\, 2^{6}}{3^{9}} \,\, = \,\, 0.27 $$

```r
x <- (choose(9,3) * 2**6)/(3**9)
x
```

b) Für die Aufgabe b berechnet sich die Anzahl der günstigen Versuchausgänge folgendermaßen:
Für den 1. Wagen gibt es genau $ \left( \begin{array}{c} 9 \\ 3 \end{array} \right) $ Möglichkeiten. 
Für den 2. Wagen gibt es dann nur mehr $ \left( \begin{array}{c} 6 \\ 3 \end{array} \right) $ Möglichkeiten da 3 Personen schon im 1. Wagen sitzen. Die restlichen 3 Personen müssen dann automatisch in den 3. Wagen einsteigen.

$$\Large \frac{\left( \begin{array}{c} 9 \\ 3 \end{array}\right) \,\, \cdot \,\, \left( \begin{array}{c} 6 \\ 3 \end{array}\right)}{3^{9}} \,\, = \,\, 0.085 $$

```r
x <- (choose(9,3) * choose(6,3))/(3**9)
x
```

c) Für die Aufgabe b wird bestimmt auf viele Arten 2, 3 und 4 elementige Teilmengen aus einer 9-elementigen Menge gezogen werden können. Das Ergebnis davon wird mit 3! multipliziert, da die Angabe nicht verlangt, dass diese vorher genannten k-elementigen Mengen auf bestimmten Wagonen sind. Das 2er Paar kann im ersten, zweiten oder dritten Wagon sein. Also 3 Möglichkeiten. Das 3er Paar kann dann noch entweder im zweiten oder dritten Wagon sein. Daher 2 Möglichkeiten und für das 4er Paar bleibt dann nur noch der dritte Wagon übrig.

$$\Large \frac{\left( \begin{array}{c} 9 \\ 2 \end{array}\right) \,\, \cdot \,\, \left( \begin{array}{c} 7 \\ 3 \end{array}\right) \,\, \cdot \,\, 3!}{3^{9}} \,\, = \,\, 0.38 $$

### Aufgabe 2

Aus einer Tabelle (einstelliger) Zufallsziffern werden nacheinander Ziffern entnommen. Wie viele Ziffern muss man mindestens entnehmen, damit unter ihnen mit einer Wahrscheinlichkeit $\geq 0.95$ mindestens eine ungerade ist.

Eine Tabelle einstelliger Zufallsziffern kann maximal 10 einstellige Ziffern haben (0 - 9). 5 der Ziffern sind ungerade und daher ist die Wahrscheinlichkeit eine ungerade Ziffer zu ziehen gleich $\frac{1}{2}$. Für die Wahrscheinlich beim zweiten Zug wieder eine ungerade Ziffer zu ziehen erhalten wir $\left(\frac{1}{2}\right)^2$. Die Wahrscheinlichkeit nach n Zügen zumindest 1 ungerade Ziffer zu ziehen ist daher $\left(\frac{1}{2}\right)^n$.

Unsere Wahrscheinlichkeit soll größer als 0.95 sein also lautet unsere Gleichung:

$$\Large \left(\frac{1}{2}\right)^{n} \geq 0.95  $$

Mit Hilfe des natürlich Logarithmus und Umformen der Gleichung erhalten wir einen Ausdruck für n.

$$\Large n \,\, \geq \,\, \frac{ln(0.95)}{ln\left(\frac{1}{2}\right)} $$

```r
log(0.95)/log(1/2)
```

$$\Large n \,\, \geq \,\, 0.07 $$
